```python
import numpy as np
```

```python
def convert_to(a: np.ndarray):
    return np.where(a <= 0, -1, 1)

def convert_back(a: np.ndarray):
    return np.where(a > 0, 1, 0)
```

```python
def get_binary(width):
    values = []
    for i in range(2**width):
        str_repr = list(np.binary_repr(i, width))
        str_repr = ','.join(str_repr)
        values.append(np.fromstring(str_repr, dtype=int, sep=','))
    return values
```

```python
def train(training_set: list[tuple[np.ndarray, np.ndarray]]):
    sets = training_set[0]
    inputs, outputs = sets
    model = np.zeros([inputs.size, outputs.size])
    for input, output in training_set:
        # convert inputs / outputs to -1 / 1 form
        # for inputs, reshape (transpose in this case, but somehow transpose won't work)
        input_conv = convert_to(np.reshape(input, shape=(input.size, 1)))
        output_conv = convert_to(np.reshape(output, shape=(1, output.size)))

        delta_w = np.matmul(input_conv, output_conv)
        model = np.add(model,delta_w)

    return model
```
